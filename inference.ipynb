```python
from tensorflow.keras.models import load_model
import joblib

# Load deep learning models
modelLSTM = load_model('models/lstm_model.keras')
modelGRU = load_model('models/gru_model.keras')
modelCNN = load_model('models/cnn_model.keras')

# Load machine learning models
svm_model = joblib.load('models/svm_model.pkl')
logreg_model = joblib.load('models/logreg_model.pkl')

# Load vectorizers
tokenizer = joblib.load('models/tokenizer.pkl')
vectorizer_bow = joblib.load('models/bow_vectorizer.pkl')
vectorizer_tfidf = joblib.load('models/tfidf_vectorizer.pkl')
```

```python
sample_texts = [
    "Raiden’s theme and animation gives me goosebumps every time!",
    "This demo is on a whole different level. Truly majestic!",
    "I never get tired watching this. Absolute perfection.",
    "Her theme slaps so hard, I get chills every single time.",
    "I miss when Genshin trailers had this level of quality..."
]
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# ML vector
X_tfidf = vectorizer_tfidf.transform(sample_texts)  # for LogReg
X_bow = vectorizer_bow.transform(sample_texts)      # for SVM

# DL vector
sequences = tokenizer.texts_to_sequences(sample_texts)
padded_sequences = pad_sequences(sequences, maxlen=200)
```

```python
import numpy as np

def decode_prediction(pred):
    label_map = {0: "Negatif", 1: "Netral", 2: "Positif"}
    return label_map[int(pred)]

print("=== LogReg Predictions ===")
logreg_preds = logreg_model.predict(X_tfidf)
for text, pred in zip(sample_texts, logreg_preds):
    print(f"{text} → {decode_prediction(pred)}")

print("\n=== SVM Predictions ===")
svm_preds = svm_model.predict(X_bow)
for text, pred in zip(sample_texts, svm_preds):
    print(f"{text} → {decode_prediction(pred)}")

print("\n=== LSTM Predictions ===")
lstm_preds = np.argmax(modelLSTM.predict(padded_sequences), axis=1)
for text, pred in zip(sample_texts, lstm_preds):
    print(f"{text} → {decode_prediction(pred)}")

print("\n=== GRU Predictions ===")
gru_preds = np.argmax(modelGRU.predict(padded_sequences), axis=1)
for text, pred in zip(sample_texts, gru_preds):
    print(f"{text} → {decode_prediction(pred)}")

print("\n=== CNN Predictions ===")
cnn_preds = np.argmax(modelCNN.predict(padded_sequences), axis=1)
for text, pred in zip(sample_texts, cnn_preds):
    print(f"{text} → {decode_prediction(pred)}")
```
